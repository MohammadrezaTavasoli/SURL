```python
import json
import sys
from pathlib import Path

import torch

sys.path.append("/content")
import srul_cifar_next_experiment as exp


# ------------------------------------------------------------------
# Load configuration, data, and trained autoencoder
# ------------------------------------------------------------------

RUN_DIR = Path(
    "/content/drive/MyDrive/SRUL_CIFAR10/quality_v1/seed_0"
)

with open(RUN_DIR / "config.json", "r") as f:
    cfg = json.load(f)

device = exp.get_device()
print("Device:", device)

train_set, test_set = exp.make_datasets(
    name="cifar10",
    root="/content/data",
    train_samples=None,
    test_samples=None,
    seed=0,
)

_, test_loader = exp.make_loaders(
    train_set=train_set,
    test_set=test_set,
    batch_size=256,
    num_workers=2,
    device=device,
)


def load_state_dict_safely(path):
    obj = torch.load(path, map_location=device)
    # Supports both a final state_dict and a resumable checkpoint.
    return obj["model"] if isinstance(obj, dict) and "model" in obj else obj


ae = exp.SphericalAutoencoder(
    latent_dim=cfg["latent_dim"],
    base_channels=cfg["base_channels"],
).to(device)

ae.load_state_dict(
    load_state_dict_safely(
        RUN_DIR / "checkpoints" / "autoencoder_final.pt"
    )
)
ae.eval()

print("Autoencoder loaded.")
```

```python
# ------------------------------------------------------------------
# Collect real images and their reconstructions
# ------------------------------------------------------------------

max_samples = 5000
real_images = []
reconstructed_images = []
count = 0

with torch.no_grad():
    for images, _ in test_loader:
        images = images.to(device)

        z = ae.encode(images)
        recon = ae.decode(z)

        remaining = max_samples - count
        images = images[:remaining]
        recon = recon[:remaining]

        real_images.append(exp.to_uint8(images).cpu())
        reconstructed_images.append(exp.to_uint8(recon).cpu())

        count += images.size(0)
        if count >= max_samples:
            break

real_uint8 = torch.cat(real_images, dim=0)
recon_uint8 = torch.cat(reconstructed_images, dim=0)

print("Collected:", real_uint8.shape[0], "images")
```

```python
# ------------------------------------------------------------------
# Image-feature comparison: real vs reconstruction
# ------------------------------------------------------------------

real_features = exp.extract_resnet_features(
    real_uint8,
    device=device,
    batch_size=128,
)

recon_features = exp.extract_resnet_features(
    recon_uint8,
    device=device,
    batch_size=128,
)

reconstruction_pr = exp.feature_precision_recall(
    real_features=real_features,
    fake_features=recon_features,
    device=device,
    nearest_k=5,
    chunk_size=256,
)

reconstruction_fid_kid = exp.compute_fid_kid(
    real_uint8=real_uint8,
    fake_uint8=recon_uint8,
    device=device,
    batch_size=128,
)

print("Real vs reconstruction precision/recall:")
print(reconstruction_pr)

print("\nReal vs reconstruction FID/KID:")
print(reconstruction_fid_kid)
```

```python
# ------------------------------------------------------------------
# Load the trained prior networks
# ------------------------------------------------------------------

def load_prior(method):
    field = exp.LatentVectorField(
        latent_dim=cfg["latent_dim"],
        width=cfg["prior_width"],
        depth=cfg["prior_depth"],
        time_dim=cfg["time_dim"],
    ).to(device)

    field.load_state_dict(
        load_state_dict_safely(
            RUN_DIR / "checkpoints" / f"{method}_final.pt"
        )
    )
    field.eval()
    return field


priors = {
    method: load_prior(method)
    for method in ["chord", "rfm", "srul"]
}

print("Prior checkpoints loaded.")
```

```python
# ------------------------------------------------------------------
# Encode real test images
# ------------------------------------------------------------------

max_samples = 5000
real_latents = []
count = 0

with torch.no_grad():
    for images, _ in test_loader:
        images = images.to(device)

        z = ae.encode(images)

        remaining = max_samples - count
        z = z[:remaining]

        real_latents.append(z.cpu())
        count += z.size(0)

        if count >= max_samples:
            break

real_latents = torch.cat(real_latents, dim=0)

print("Real latent shape:", real_latents.shape)
print(
    "Real latent norm:",
    real_latents.norm(dim=1).mean().item()
)
```

```python
# ------------------------------------------------------------------
# Sample final latents from each flow prior
# ------------------------------------------------------------------

@torch.no_grad()
def sample_final_latents(
    method,
    field,
    num_samples=5000,
    steps=100,
    batch_size=256,
):
    all_latents = []

    for start in range(0, num_samples, batch_size):
        batch = min(batch_size, num_samples - start)

        z = exp.sample_uniform_sphere(
            batch,
            ae.latent_dim,
            device,
        )

        for index in range(steps, 0, -1):
            t = torch.full(
                (batch,),
                index / steps,
                device=device,
            )

            raw_velocity = field(z, t)
            dt = -1.0 / steps

            if method == "chord":
                z = z + dt * raw_velocity
            elif method in {"rfm", "srul"}:
                tangent_velocity = exp.tangent_projection(
                    z,
                    raw_velocity,
                )
                z = exp.exp_map_sphere(
                    z,
                    dt * tangent_velocity,
                )
            else:
                raise ValueError(method)

        # The decoder receives normalized chord latents in the experiment.
        if method == "chord":
            z = exp.normalize_to_sphere(z)

        all_latents.append(z.cpu())

    return torch.cat(all_latents, dim=0)


generated_latents = {
    method: sample_final_latents(
        method,
        priors[method],
        num_samples=5000,
        steps=100,
    )
    for method in ["chord", "rfm", "srul"]
}
```

```python
# ------------------------------------------------------------------
# Latent-space precision and recall
# ------------------------------------------------------------------

latent_results = {}

for method, fake_latents in generated_latents.items():
    result = exp.feature_precision_recall(
        real_features=real_latents,
        fake_features=fake_latents,
        device=device,
        nearest_k=5,
        chunk_size=256,
    )

    latent_results[method] = result
    print(method, result)
```

```python
import torch

num_samples = 5000
batch_size = 256

# -------------------------------------------------------------
# Sample directly from the uniform sphere, without any prior
# -------------------------------------------------------------
uniform_latents = exp.sample_uniform_sphere(
    num_samples,
    ae.latent_dim,
    device,
).cpu()

uniform_latent_pr = exp.feature_precision_recall(
    real_features=real_latents,
    fake_features=uniform_latents,
    device=device,
    nearest_k=5,
    chunk_size=256,
)

print("Real latent vs uniform sphere:")
print(uniform_latent_pr)
```

```python
uniform_images = []

with torch.no_grad():
    for start in range(0, num_samples, batch_size):
        z = uniform_latents[start:start + batch_size].to(device)
        x_hat = ae.decode(z)
        uniform_images.append(exp.to_uint8(x_hat).cpu())

uniform_images = torch.cat(uniform_images, dim=0)
```

```python
uniform_features = exp.extract_resnet_features(
    uniform_images,
    device=device,
    batch_size=128,
)

uniform_image_pr = exp.feature_precision_recall(
    real_features=real_features,
    fake_features=uniform_features,
    device=device,
    nearest_k=5,
    chunk_size=256,
)

uniform_fid_kid = exp.compute_fid_kid(
    real_uint8=real_uint8[:num_samples],
    fake_uint8=uniform_images,
    device=device,
    batch_size=128,
)

print("Real image vs decoded uniform sphere:")
print(uniform_image_pr)
print(uniform_fid_kid)
```

```python
def normalize_tokens(h: torch.Tensor, eps: float = 1e-8) -> torch.Tensor:
    """
    h: [batch, channels, height, width]

    Every spatial token is normalized across its channel dimension.
    """
    norm = h.pow(2).sum(dim=1, keepdim=True).sqrt()
    return h / norm.clamp_min(eps)
```

```python
def tangent_projection_tokens(
    z: torch.Tensor,
    u: torch.Tensor,
) -> torch.Tensor:
    """
    z, u: [batch, channels, height, width]
    """
    radial_coefficient = (u * z).sum(dim=1, keepdim=True)
    return u - radial_coefficient * z
```

```python
def exp_map_tokens(
    z: torch.Tensor,
    v: torch.Tensor,
    eps: float = 1e-8,
) -> torch.Tensor:
    """
    Applies the sphere exponential map independently to each token.
    """
    v_norm = v.pow(2).sum(dim=1, keepdim=True).sqrt()
    direction = v / v_norm.clamp_min(eps)

    mapped = (
        torch.cos(v_norm) * z
        + torch.sin(v_norm) * direction
    )

    return normalize_tokens(mapped)
```

```python
from pathlib import Path
import json

import pandas as pd
from IPython.display import display


ROOT = Path("/content/drive/MyDrive/SRUL_CIFAR10/spatial_v1")
SEEDS = [0, 1, 2]

# ------------------------------------------------------------
# Collect generation and geometry metrics
# ------------------------------------------------------------

generation_frames = []

for seed in SEEDS:
    path = ROOT / f"seed_{seed}" / "generation_metrics.csv"

    if not path.exists():
        print(f"Missing: {path}")
        continue

    frame = pd.read_csv(path)
    frame["seed"] = seed
    generation_frames.append(frame)

if not generation_frames:
    raise RuntimeError("No completed generation_metrics.csv files were found.")

all_results = pd.concat(generation_frames, ignore_index=True)

metrics = [
    "fid",
    "kid_mean",
    "feature_precision",
    "feature_recall",
    "latent_feature_precision",
    "latent_feature_recall",
    "min_mean_path_norm",
    "mean_abs_path_norm_error",
    "mean_raw_radial_fraction",
]

available_metrics = [m for m in metrics if m in all_results.columns]

summary = (
    all_results
    .groupby("method")[available_metrics]
    .agg(["mean", "std"])
    .sort_values(("fid", "mean"))
)

print("All individual runs:")
display(
    all_results[
        ["method", "seed"] + available_metrics
    ].sort_values(["method", "seed"])
)

print("\nMean ± standard deviation across seeds:")
display(summary)

all_results.to_csv(
    ROOT / "all_seed_results.csv",
    index=False,
)

summary.to_csv(
    ROOT / "three_seed_summary.csv"
)

print("\nSaved:")
print(ROOT / "all_seed_results.csv")
print(ROOT / "three_seed_summary.csv")
```

```python
def mean_std_text(group: pd.DataFrame, metric: str, digits: int = 2) -> str:
    mean = group[metric].mean()
    std = group[metric].std(ddof=1)
    return f"{mean:.{digits}f} ± {std:.{digits}f}"


rows = []

for method, group in all_results.groupby("method"):
    rows.append({
        "Method": method,
        "FID ↓": mean_std_text(group, "fid", 2),
        "KID ↓": mean_std_text(group, "kid_mean", 4),
        "Precision ↑": mean_std_text(group, "feature_precision", 4),
        "Recall ↑": mean_std_text(group, "feature_recall", 4),
        "Latent precision ↑": mean_std_text(
            group, "latent_feature_precision", 4
        ),
        "Latent recall ↑": mean_std_text(
            group, "latent_feature_recall", 4
        ),
    })

paper_table = pd.DataFrame(rows).sort_values("Method")
display(paper_table)

paper_table.to_csv(
    ROOT / "paper_result_table.csv",
    index=False,
)
```
